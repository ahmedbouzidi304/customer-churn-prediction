**Business problem.** Acquiring a new telecom subscriber costs far more than keeping one. The retention team can only contact a limited number of customers, so they need a ranked list of *who is likely to leave*, an explanation of *what drives the risk*, and a clear rule for *where to draw the line* so the campaign makes money.

**What this notebook delivers**
- a leak-free `scikit-learn` pipeline (cleaning → feature engineering → preprocessing → model);
- a cross-validated comparison of 5 models and Optuna-tuned LightGBM;
- a decision threshold chosen by **expected campaign profit**, not by the default 0.5;
- global and per-customer explanations with **SHAP**;
- a saved, reusable model artifact with a `predict_churn()` function.

<div style="display:flex;align-items:center;gap:10px;margin:18px 0 8px 0;border-left:4px solid #3DD6C6;padding-left:10px"><span style="background:#1B1530;color:#3DD6C6;font-family:Consolas,'Courier New',monospace;font-size:12px;font-weight:bold;padding:2px 8px;border-radius:10px">↳ 2.1</span><span style="font-family:'Segoe UI',Arial,Helvetica,sans-serif;font-size:17px;font-weight:700;color:#1B1530">Cleaning rules</span></div>

`TotalCharges` is stored as text because brand-new customers (tenure = 0) have a blank value: they simply have not been billed yet, so the correct value is **0**, not a statistical imputation. `SeniorCitizen` is coded 0/1 while every other yes/no column is text, so it is harmonised. The target is mapped to 0/1.

```python
def clean(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].astype(str).str.strip().replace("", np.nan))
    new_customer = df["TotalCharges"].isna() & (df["tenure"] == 0)
    df.loc[new_customer, "TotalCharges"] = 0.0
    df["SeniorCitizen"] = df["SeniorCitizen"].map({0: "No", 1: "Yes"})
    if "Churn" in df:
        df["Churn"] = (df["Churn"] == "Yes").astype(int)
    return df.set_index("customerID")

df = clean(raw)
print("Blank TotalCharges fixed:", (raw["TotalCharges"].str.strip() == "").sum())
print("Remaining missing values:", int(df.isna().sum().sum()))
df.info()
```

Every feature below only uses the customer's own row, so it can be computed before the train/test split without leaking information.

| Feature | Idea |
|---|---|
| `n_services` | number of add-on services: more services → more switching friction |
| `avg_monthly_spend` | `TotalCharges / tenure`: historical average bill |
| `charge_increase` | current bill minus historical average: recent price rises |
| `is_month_to_month` | the strongest single risk signal from EDA |
| `auto_pay` | automatic payment methods reduce passive churn |
| `has_family` | partner or dependents |
| `no_protection` | internet user without security *and* tech support |
| `tenure_bucket` | lifecycle stage |

```python
SERVICES = ["PhoneService", "MultipleLines", "OnlineSecurity", "OnlineBackup",
            "DeviceProtection", "TechSupport", "StreamingTV", "StreamingMovies"]

def add_features(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out["n_services"] = (out[SERVICES] == "Yes").sum(axis=1)
    out["avg_monthly_spend"] = np.where(out["tenure"] > 0, out["TotalCharges"] / out["tenure"].clip(lower=1),
                                        out["MonthlyCharges"])
    out["charge_increase"] = out["MonthlyCharges"] - out["avg_monthly_spend"]
    out["is_month_to_month"] = (out["Contract"] == "Month-to-month").astype(int)
    out["auto_pay"] = out["PaymentMethod"].str.contains("automatic").astype(int)
    out["has_family"] = ((out["Partner"] == "Yes") | (out["Dependents"] == "Yes")).astype(int)
    out["no_protection"] = ((out["InternetService"] != "No")
                            & (out[["OnlineSecurity", "TechSupport"]] == "No").all(axis=1)).astype(int)
    out["tenure_bucket"] = pd.cut(out["tenure"], bins=[-1, 6, 12, 24, 48, 72],
                                  labels=["0-6m", "7-12m", "13-24m", "25-48m", "49-72m"]).astype(str)
    return out

X = add_features(df.drop(columns="Churn"))
y = df["Churn"]
new_cols = ["n_services", "avg_monthly_spend", "charge_increase", "is_month_to_month",
            "auto_pay", "has_family", "no_protection", "tenure_bucket"]
X[new_cols].head()
```

20% of customers are locked away as a **test set** that is only used once, at the end. All imputation, scaling and one-hot encoding live *inside* the pipeline, so they are re-fitted on each training fold during cross-validation.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=SEED)
num_cols = X.select_dtypes("number").columns.tolist()
cat_cols = X.select_dtypes(exclude="number").columns.tolist()

def make_preprocessor() -> ColumnTransformer:
    numeric = Pipeline([("impute", SimpleImputer(strategy="median")), ("scale", StandardScaler())])
    categorical = Pipeline([("impute", SimpleImputer(strategy="most_frequent")),
                            ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))])
    return ColumnTransformer([("num", numeric, num_cols), ("cat", categorical, cat_cols)],
                             verbose_feature_names_out=False)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
print(f"train: {X_train.shape}  churn={y_train.mean():.1%}")
print(f"test : {X_test.shape}  churn={y_test.mean():.1%}")
print(f"{len(num_cols)} numeric + {len(cat_cols)} categorical features")
```

Five models, same pipeline, same 5 folds. Class imbalance is handled with class weights rather than resampling, which keeps the pipeline simple.

```python
pos_weight = (y_train == 0).sum() / (y_train == 1).sum()
models = {
    "Dummy (prior)": DummyClassifier(strategy="prior"),
    "Logistic Regression": LogisticRegression(max_iter=2000, C=0.5, class_weight="balanced"),
    "Random Forest": RandomForestClassifier(n_estimators=400, min_samples_leaf=5, class_weight="balanced",
                                            n_jobs=-1, random_state=SEED),
    "XGBoost": XGBClassifier(n_estimators=400, learning_rate=0.05, max_depth=4, subsample=0.8,
                             colsample_bytree=0.8, scale_pos_weight=pos_weight, eval_metric="logloss",
                             n_jobs=-1, random_state=SEED),
    "LightGBM": LGBMClassifier(n_estimators=400, learning_rate=0.05, num_leaves=31, subsample=0.8,
                               subsample_freq=1, colsample_bytree=0.8, class_weight="balanced",
                               random_state=SEED, verbose=-1),
}
scoring = {"roc_auc": "roc_auc", "pr_auc": "average_precision", "f1": "f1"}

rows = []
for name, model in models.items():
    pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
    s = cross_validate(pipe, X_train, y_train, cv=cv, scoring=scoring)
    rows.append({"model": name, "ROC-AUC": s["test_roc_auc"].mean(), "ROC-AUC std": s["test_roc_auc"].std(),
                 "PR-AUC": s["test_pr_auc"].mean(), "F1 @0.5": s["test_f1"].mean(),
                 "fit time (s)": s["fit_time"].mean()})
cv_results = pd.DataFrame(rows).set_index("model").sort_values("ROC-AUC", ascending=False)
cv_results.style.format("{:.3f}").background_gradient(subset=["ROC-AUC", "PR-AUC"], cmap="Blues")
```

Optuna's TPE sampler proposes new hyperparameters based on previous trials. Each trial is scored with the same 5-fold CV ROC-AUC, on the training set only.

```python
def objective(trial: optuna.Trial) -> float:
    params = dict(
        n_estimators=trial.suggest_int("n_estimators", 100, 800, step=50),
        learning_rate=trial.suggest_float("learning_rate", 0.01, 0.2, log=True),
        num_leaves=trial.suggest_int("num_leaves", 8, 64),
        max_depth=trial.suggest_int("max_depth", 3, 8),
        min_child_samples=trial.suggest_int("min_child_samples", 10, 120),
        subsample=trial.suggest_float("subsample", 0.6, 1.0),
        colsample_bytree=trial.suggest_float("colsample_bytree", 0.4, 1.0),
        reg_alpha=trial.suggest_float("reg_alpha", 1e-3, 10, log=True),
        reg_lambda=trial.suggest_float("reg_lambda", 1e-3, 10, log=True),
    )
    model = LGBMClassifier(**params, subsample_freq=1, class_weight="balanced", random_state=SEED, verbose=-1)
    pipe = Pipeline([("prep", make_preprocessor()), ("model", model)])
    return cross_val_score(pipe, X_train, y_train, cv=cv, scoring="roc_auc").mean()

N_TRIALS = 40
t0 = time.time()
study = optuna.create_study(direction="maximize", sampler=optuna.samplers.TPESampler(seed=SEED))
study.optimize(objective, n_trials=N_TRIALS)
print(f"{N_TRIALS} trials in {time.time() - t0:.0f}s | best CV ROC-AUC = {study.best_value:.4f}")
pd.Series(study.best_params).to_frame("best value")
```

Three candidates go forward: the strong linear baseline, the tuned LightGBM, and a **soft-voting ensemble** of both (it averages their probabilities, combining linear and non-linear signal). The one with the best CV ROC-AUC becomes the production model.

```python
lgbm_tuned = LGBMClassifier(**study.best_params, subsample_freq=1, class_weight="balanced",
                            random_state=SEED, verbose=-1)
logreg = LogisticRegression(max_iter=2000, C=0.5, class_weight="balanced")
candidates = {
    "Logistic Regression": logreg,
    "LightGBM (tuned)": lgbm_tuned,
    "Ensemble LR + LGBM": VotingClassifier([("lr", logreg), ("lgbm", lgbm_tuned)], voting="soft"),
}
pipes = {name: Pipeline([("prep", make_preprocessor()), ("model", m)]) for name, m in candidates.items()}
cand_cv = pd.Series({name: cross_val_score(p, X_train, y_train, cv=cv, scoring="roc_auc").mean()
                     for name, p in pipes.items()}, name="CV ROC-AUC").sort_values(ascending=False)
FINAL_NAME = cand_cv.idxmax()
print("Selected model:", FINAL_NAME)
cand_cv.to_frame().style.format("{:.4f}")
```

A retention campaign contacts every customer whose score is above a threshold. Assumptions (explicit, and easy to change):

| Parameter | Value | Meaning |
|---|---|---|
| `OFFER_COST` | \$50 | cost of one retention offer (discount + agent time) |
| `SUCCESS_RATE` | 30% | share of contacted *would-be churners* who stay |
| `HORIZON_MONTHS` | 12 | a saved customer is worth 12 months of their current bill |

To avoid tuning on the test set, the threshold is chosen on **out-of-fold predictions on the training set**, then applied once to the test set.

```python
OFFER_COST, SUCCESS_RATE, HORIZON_MONTHS = 50.0, 0.30, 12

def campaign_profit(p, y_true, monthly, threshold):
    target = p >= threshold
    saved_value = SUCCESS_RATE * HORIZON_MONTHS * monthly
    return float(((target & (y_true == 1)) * saved_value).sum() - target.sum() * OFFER_COST)

thresholds = np.round(np.arange(0.02, 0.99, 0.01), 2)
oof = cross_val_predict(final_model, X_train, y_train, cv=cv, method="predict_proba")[:, 1]
oof_profit = [campaign_profit(oof, y_train.values, X_train["MonthlyCharges"].values, t) for t in thresholds]
BEST_T = float(thresholds[int(np.argmax(oof_profit))])

m_test, y_t = X_test["MonthlyCharges"].values, y_test.values
test_profit = [campaign_profit(p_test, y_t, m_test, t) for t in thresholds]
policies = pd.DataFrame({
    "customers contacted": [0, len(y_t), int((p_test >= 0.5).sum()), int((p_test >= BEST_T).sum())],
    "test profit ($)": [0.0, campaign_profit(p_test, y_t, m_test, 0.0),
                        campaign_profit(p_test, y_t, m_test, 0.5), campaign_profit(p_test, y_t, m_test, BEST_T)],
}, index=["No campaign", "Contact everyone", "Default threshold 0.5", f"Profit-optimal threshold {BEST_T:.2f}"])
print(f"Threshold chosen on out-of-fold train predictions: {BEST_T:.2f}")
policies.style.format({"test profit ($)": "{:,.0f}", "customers contacted": "{:,}"})
```

```python
fig, ax = plt.subplots(figsize=(10, 3.8))
ax.plot(thresholds, np.array(test_profit) / 1000, color=C_STAY, lw=2)
ax.axvline(BEST_T, color=C_CHURN, lw=2)
ax.axvline(0.5, color=MUTED, ls="--", lw=1)
ax.text(BEST_T, max(test_profit) / 1000, f"  chosen t = {BEST_T:.2f}", color=TEXT, va="bottom")
ax.text(0.5, min(test_profit) / 1000, "  default 0.5", color=MUTED, va="bottom", fontsize=9)
ax.axhline(0, color=TEXT, lw=0.8)
ax.set(xlabel="decision threshold", ylabel="profit on test set ($k)", title="Campaign profit vs threshold (test set)")
plt.show()

cm = confusion_matrix(y_test, (p_test >= BEST_T).astype(int))
cm_df = pd.DataFrame(cm, index=["actual: stayed", "actual: churned"], columns=["not contacted", "contacted"])
print(f"Recall on churners at t={BEST_T:.2f}: {cm[1, 1] / cm[1].sum():.1%} | precision: {cm[1, 1] / cm[:, 1].sum():.1%}")
cm_df
```

If the budget is fixed (e.g. *call 20% of customers*), the ranking quality is what matters. Decile 1 = the 10% highest-risk customers.

```python
dec = pd.DataFrame({"p": p_test, "y": y_t})
dec["decile"] = pd.qcut(dec["p"].rank(method="first", ascending=False), 10, labels=range(1, 11))
lift = dec.groupby("decile", observed=True).agg(churn_rate=("y", "mean"), customers=("y", "size"), churners=("y", "sum"))
lift["lift"] = lift["churn_rate"] / y_t.mean()
lift["cumulative % of churners"] = lift["churners"].cumsum() / lift["churners"].sum()

fig, ax = plt.subplots(figsize=(10, 3.6))
ax.bar(lift.index.astype(str), lift["churn_rate"], color=C_CHURN, width=0.65)
ax.axhline(y_t.mean(), color=MUTED, ls="--", lw=1)
for i, (r, lf) in enumerate(zip(lift["churn_rate"], lift["lift"])):
    ax.text(i, r + 0.01, f"{lf:.1f}x", ha="center", color=TEXT, fontsize=9)
ax.yaxis.set_major_formatter(PercentFormatter(1.0)); ax.grid(axis="x", visible=False)
ax.set(xlabel="risk decile (1 = highest predicted risk)", title="Observed churn rate per decile (labels = lift vs average)")
plt.show()
print(f"Top 20% riskiest customers contain {lift['cumulative % of churners'].iloc[1]:.0%} of all churners.")
lift.style.format({"churn_rate": "{:.1%}", "lift": "{:.2f}", "cumulative % of churners": "{:.1%}"})
```

SHAP values split each prediction into additive contributions of the features. Positive values push towards churn. Numeric features are shown in their original units.

```python
prep, clf = final_model.named_steps["prep"], final_model.named_steps["model"]
feat_names = list(prep.get_feature_names_out())
explain_idx = X_test.sample(n=min(500, len(X_test)), random_state=SEED).index
Xt = pd.DataFrame(prep.transform(X_test.loc[explain_idx]), columns=feat_names, index=explain_idx)
Xt_bg = pd.DataFrame(prep.transform(X_train.sample(200, random_state=SEED)), columns=feat_names)

if isinstance(clf, LGBMClassifier):
    sv = shap.TreeExplainer(clf)(Xt)
elif isinstance(clf, LogisticRegression):
    sv = shap.LinearExplainer(clf, Xt_bg)(Xt)
else:  # ensemble: model-agnostic permutation explainer on the probability
    f = lambda a: clf.predict_proba(pd.DataFrame(a, columns=feat_names))[:, 1]
    sv = shap.Explainer(f, shap.maskers.Independent(Xt_bg, max_samples=100), algorithm="permutation")(Xt)
if sv.values.ndim == 3:
    sv = sv[:, :, 1]

# show numeric features in original units
scaler = prep.named_transformers_["num"].named_steps["scale"]
Xt_display = Xt.copy()
Xt_display[num_cols] = Xt[num_cols].values * scaler.scale_ + scaler.mean_
sv = shap.Explanation(values=sv.values, base_values=sv.base_values, data=Xt_display.values, feature_names=feat_names)
print(f"Explained {len(Xt)} test customers with {type(clf).__name__}")
```

```python
def predict_churn(raw_customers: pd.DataFrame, model_path=ARTIFACTS / "churn_model.joblib",
                  threshold: float = BEST_T) -> pd.DataFrame:
    """Score raw customer records (same columns as the source CSV, without Churn)."""
    model = joblib.load(model_path)
    feats = add_features(clean(raw_customers))
    proba = model.predict_proba(feats)[:, 1]
    return pd.DataFrame({"churn_probability": proba.round(3), "contact": proba >= threshold},
                        index=feats.index).sort_values("churn_probability", ascending=False)

new_customers = raw.drop(columns="Churn").sample(8, random_state=7)
predict_churn(new_customers)
```
